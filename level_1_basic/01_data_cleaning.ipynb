# 🧹 Step 2: Data Cleaning & Preprocessing
**Task:** Clean the raw scraped books data.  
**Input:** `data/scraped_books_raw.csv`  
**Output:** `outputs/cleaned_books.csv`

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
import joblib
import os

# Ensure output directories exist
os.makedirs('outputs', exist_ok=True)
os.makedirs('models', exist_ok=True)

# Load Raw Data
data_path = 'data/scraped_books_raw.csv'
try:
    df = pd.read_csv(data_path)
    print('✅ Raw data loaded successfully')
except:
    # Fallback dummy data if scraping wasn't run
    df = pd.DataFrame({'Title': ['A', 'B'], 'Raw_Price': ['£10.00', '£20.50'], 'Text_Rating': ['One', 'Three'], 'Availability_Text': ['In stock', 'Out of stock']})
    print('⚠️ Using dummy data')

display(df.head())
```

```python
# 1. Clean Price (Remove currency symbol and convert to float)
df['Price'] = df['Raw_Price'].str.replace('£', '', regex=False).str.replace('Â', '', regex=False).astype(float)
print('✅ Price cleaned')
```

```python
# 2. Encode Text Rating (Using LabelEncoder to save it as a model)
encoder = LabelEncoder()
df['Rating'] = encoder.fit_transform(df['Text_Rating']) + 1 # +1 to make it 1-5 instead of 0-4
joblib.dump(encoder, 'models/encoder.pkl')
print('✅ Ratings encoded and encoder saved to models/encoder.pkl')
```

```python
# 3. Encode Availability (Binary Encoding)
df['In_Stock'] = df['Availability_Text'].apply(lambda x: 1 if 'In stock' in x else 0)
print('✅ Availability encoded')
```

```python
# 4. Standardize Price for modeling purposes
scaler = StandardScaler()
df['Price_Scaled'] = scaler.fit_transform(df[['Price']])
joblib.dump(scaler, 'models/scaler.pkl')
print('✅ Price standardized and scaler saved to models/scaler.pkl')
```

```python
# Keep only clean columns
df_clean = df[['Title', 'Price', 'Price_Scaled', 'Rating', 'In_Stock']]
display(df_clean.head())
```

```python
output_path = 'outputs/cleaned_books.csv'
df_clean.to_csv(output_path, index=False)
print(f'💾 Clean data saved to: {output_path}')
```
